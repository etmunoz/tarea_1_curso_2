## 3 · Costear para 4 personas

Costo por ingrediente (cantidad × precio por su unidad), total y por persona (CLP).

```python
c = costear_receta.invoke({'nombre':'Pastel de choclo','personas':4})
for d in c['detalle']:
    print(f"  {d['ingrediente']:16} {d['cantidad']:>6} {d['unidad']:7} x {d['precio_unitario_clp']:>5} = {d['costo_clp']:>6} CLP")
print(f"TOTAL: {c['total_clp']} CLP  ·  por persona: {c['por_persona_clp']} CLP")
```
