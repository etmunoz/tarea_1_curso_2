## 3 · La ronda de preguntas

Mismo hilo: el supervisor elige el agente, consulta y devuelve `[Nombre] …`. La última pregunta no es de nadie.

```python
from langchain_core.messages import HumanMessage
hilo={'configurable':{'thread_id':'nb-coord-1'}}
def turno(t):
    r=graph.invoke({'messages':[HumanMessage(content=t)]}, hilo)
    print(f"> {t}\n  ruta: {' -> '.join(r['ruta'])}\n  {r['respuesta']}\n")
turno('¿Qué técnica uso si tengo desbalance de clases?')
turno('Quiero un vuelo de Santiago a Lima el 2026-11-10')
turno('¿Quién ganó el mundial de 1962?')
```
