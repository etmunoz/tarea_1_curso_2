## 3 · Conversación con handoffs

Un mismo hilo: el agente activo cambia según la pregunta (vuelos → hoteles → actividades).

```python
hilo={'configurable':{'thread_id':'nb-swarm-1'}}
def turno(t):
    r=graph.invoke({'messages':[{'role':'user','content':t}]}, hilo)
    print(f"> {t}\n[{r.get('active_agent','vuelos')}] {texto(r['messages'][-1])}\n")
turno('Quiero un vuelo de Santiago a Lima el 2026-11-10')
turno('Y un hotel en Lima por 3 noches')
turno('¿Qué actividades hay en Lima?')
```
